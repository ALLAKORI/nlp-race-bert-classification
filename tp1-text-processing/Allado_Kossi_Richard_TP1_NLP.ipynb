```python
import re
import os
import json
```

# fonction pour la segmentation en phrases

```python
def segment(text):
    return re.findall(r"[A-Z][^.!?]*[.!?]", text)
```

# Fonction de tokenisation d'un chaine en tokens

```python
def tokenize(text):
    return re.findall(r"\w+|[^\w\s]",text)
```

# Fonction de lecture de repertoire

```python
def process_dir(dir_name:str,stats_file_name:str):
    n_file=0
    stats=[]
    for filename in os.listdir(dir_name):
        if filename.endswith(".txt"):
            nbSents = 0
            nbToks = 0
            file_path=os.path.join(dir_name,filename)
            with open(file_path,"r") as f:
                file=f.read()
                para_row=file.split("\n\n")
                nbChars=len(file)
                paragraphs = []
                for para in para_row:
                    para=para.strip()
                    phrase_list=segment(para)
                    token=[tokenize(phrase) for phrase in phrase_list]
                    nbSents+=len(phrase_list)
                    nbToks+=sum(len(toks) for toks in token)
                    paragraphs.append(token)
                json_filename = filename.replace(".txt", ".seg.json")
                json_path = os.path.join(dir_name, json_filename)
                with open(json_path, "w", encoding="utf-8") as jf:
                   json.dump(paragraphs, jf, ensure_ascii=False, indent=2)
                stats.append([filename, len(paragraphs), nbSents, nbToks, nbChars])
                n_file+=1
                
    stats_path = os.path.join(dir_name, stats_file_name)
    with open(stats_path, "w", encoding="utf-8") as sf:
        sf.write("File\tnbParas\tnbSents\tnbToks\tnbChars\n")
        for row in stats:
            sf.write("\t".join(map(str, row)) + "\n")
    return n_file
```

# Test du code :

```python
dir_name="tester_NLP"
stats_file_name="stats.tsv"
print(f'Nombre de fichier traités est : {process_dir(dir_name,stats_file_name)}')
```
